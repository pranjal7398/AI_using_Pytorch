```python
import torch
print(torch.__version__)
```

```python
# tensor - they are like arrays
# tensors are data structures in python which is used to store numeric data

arr = torch.tensor([10,20,30,40])
print(arr[1] , " is at index 1")

arr2D = torch.tensor([
    [10,20,30],
    [40,50,60]
])
print(arr2D[0][1] , " is at index [0][1]")
```

```python
# adding and multiplying tensor of same dimensions (i.e 1Xn and 1Xn)
arr1 = torch.tensor([1,2,3,4])
arr2 = torch.tensor([5,6,7,8])
print("Addition =>" , arr1 + arr2)
print("Multiplication =>" ,  arr1 * arr2)
print("We can observe each element is added and multiplied to its same index value in the other tensor i.e. arr1[0] is added and multiplied to arr2[0] and so on")
```

```python
# creating a tensor using numpy array
# Note :- both will share the same memory location
import numpy as np
arr = np.array([1,2,3,4])

arr_tensor = torch.from_numpy(arr)

print("Memory locations -")
print("  (i) of numpy array " , hex(arr.ctypes.data))
print(" (ii) of tensor" ,  hex(arr_tensor.data_ptr()))
print("See both the memory locations are same")
```
